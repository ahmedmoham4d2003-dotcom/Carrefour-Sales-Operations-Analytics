```python
cr.isnull().sum()
```

```python
cr['OrderDate'] = pd.to_datetime(cr['OrderDate'], format='mixed')
```

```python
cr['month']=cr['OrderDate'].dt.month
cr['month name']=cr['OrderDate'].dt.month_name()
cr['year']=cr['OrderDate'].dt.year
cr['day']=cr['OrderDate'].dt.day
```

```python
cr['StoreName']=cr['StoreName'].str.title()
```

```python
cr['StoreName']=cr['StoreName'].str.strip()
```

```python
cr['CustomerName']=cr['CustomerName'].str.title()
```

```python
cr['CustomerName']=cr['CustomerName'].str.strip()
```

```python
cr['Category'].value_counts()
```

```python
cr['Category']=cr['Category'].str.replace('Elec','Electronics')
```

```python
cr['Category']=cr['Category'].str.replace('ELC','Electronics')
```

```python
cr['Category']=cr['Category'].str.replace('Electronicstronics','Electronics')
```

```python
cr['Category']=cr['Category'].str.replace('electronics','Electronics')
```

```python
cr['UnitPrice'] = (
    cr['UnitPrice']
    .astype(str)
    .str.replace(r'[^0-9.]', '', regex=True))
```

```python
cr['UnitPrice'] = pd.to_numeric(cr['UnitPrice'], errors='coerce')
```

```python
cr['DiscountValue'].isnull().sum()
```

```python
cr['DiscountValue'].median()
```

```python
cr['DiscountValue'] = cr['DiscountValue'].fillna(
    cr['DiscountValue'].median())
```

```python
cr['OrderStatus']=cr['OrderStatus'].str.title()
```

```python
cr['Gender']=cr['Gender'].replace('M','Male')
cr['Gender']=cr['Gender'].replace('F','Female')
```

```python
cr.to_excel('cleaned_data.xlsx', index=False)
```

```python
cr.to_csv('cleaned_data.csv', index=False)
```
